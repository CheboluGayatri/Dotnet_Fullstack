History of HTML

HTML (Hypertext Markup Language) was first developed by Tim Berners-Lee in 1991. It was designed to
be the standard language for creating web pages. The first version, HTML 1.0, was a very basic markup
language with simple elements like headers, paragraphs, and links.

Over the years, HTML evolved:

• HTML 2.0 (1995): Standardized the existing practices and added new 

features like forms.
• HTML 3.2 (1997): Introduced tables, applets, and text flow around images.

• HTML 4.01 (1999): Brought better support for multimedia, scripting, and more.

• XHTML 1.0 (2000): A reformulation of HTML 4.01 in XML, stricter and more robust.

• HTML5 (2014): Introduced new elements like <article>, <section>, native audio and
video, canvas for 2D graphics, and APIs for enhanced functionality

Understanding <div> and <span> in HTML

<div> and <span> are both HTML elements used to group and style content, but they serve
different purposes in terms of their behavior and usage.
<div> Tag

• Type: Block-level element

• Purpose: Used to group larger sections of content, creating divisions or sections in the webpage.

It is commonly used to wrap multiple elements together and can apply styling or layout changes
through CSS.

• Behavior: A <div> takes up the full width available, starting on a new line and pushing down
other content

<span> Tag

• Type: Inline element

• Purpose: Used to group and style small portions of text within a block of content. It is often
used to apply styles or scripts to a specific part of text without disrupting the flow of
surrounding content.

• Behavior: A <span> does not start on a new line and only takes up as much width as
necessary

<div> vs. <span> in HTML

Display Behavior:

• <div>: As a block-level element, it starts on a new line and takes up the full width available. It's
used to create larger containers for content, such as sections, headers, footers, or even the
layout of entire pages.

• <span>: As an inline element, it sits within the flow of surrounding text, taking up only the
width it needs. It’s typically used for styling specific parts of text or other inline content without
affecting layout.

Usage:

• <div>: Ideal for grouping large blocks of content, organizing sections of a webpage, and
applying styles or JavaScript behaviors to entire sections.

• <span>: Best used for applying styles or behaviors to small pieces of text or inline elements
within other content.
CSS Application:

• <div>: Often used with CSS for layout purposes (e.g., grid systems, flexbox), to create
structured and styled web pages.

• <span>: Used for targeting and styling specific inline elements, like changing the color or font
of a word or phrase within a paragraph.

In summary

Use <div> for grouping larger blocks of content and organizing the layout of a webpage.

Use <span> for targeting and styling small, specific parts of inline content without altering the structure
of the document.

An image map is an image with clickable areas, allowing users to click different parts of the image to
navigate to different links. Image maps are created using the <map> tag, along with the <area> tag to
define the clickable regions.

Steps to Create an Image Map:

1. Use the <img> tag with a usemap attribute that links to a <map> element.

2. Define the map using the <map> tag with a unique name attribute.

3. Create clickable areas using the <area> tag inside the <map> element.

Attributes of <area>:

• shape: Specifies the shape of the clickable area (rect, circle, or poly).

HTML-Hyperlink 

The <a> tag in HTML is used to create hyperlinks, which allow users to click and navigate to other pages,
files, or locations within the same page. The <a> tag is one of the most important tags in HTML for
linking web content.

syntax:

<a href="URL" target="target" title="title">Link Text</a>

Attributes:

• href: Specifies the URL of the page or resource the link points to. This can be an absolute URL
(link to a different website) or a relative URL (link within the same website).

• target: Specifies where to open the linked document. Common values include:

• _blank: Opens the link in a new tab or window.

• _self: Opens the link in the same window/tab (default behavior).

• _parent: Opens the link in the parent frame.

• _top: Opens the link in the full body of the window.

• title: Provides additional information about the link, typically displayed as a tooltip when the
user hovers over the link.

URL - Uniform Resource Locator

A URL (Uniform Resource Locator) is the address used to access resources on the internet. It consists of
several components


scheme://domain:port/path?query#fragment

• scheme: Specifies the protocol used (e.g., http, https, ftp).

• domain: The domain name (e.g., www.example.com) or IP address.

• port: The port number (optional, default is 80 for HTTP and 443 for HTTPS).

• path: The path to the resource on the server (e.g., /about).

• query: A string of key-value pairs, often used for parameters in web applications (e.g.,
?id=123).

• fragment: A reference to a specific section within a webpage (e.g., #section1).

<a href="URL" target="target" title="title">Link Text</a>

URL Encoding in HTML

URL encoding is a process of converting special characters in a URL into a format that can be transmitted
over the internet. URLs can only be sent over the Internet using the ASCII character set, so
characters that are not part of this set need to be encoded.

Common Encodings:

• Space: Encoded as %20

• Ampersand (&): Encoded as %26

• Question Mark (?): Encoded as %3F

• Equal Sign (=): Encoded as %3D

• Slash (/): Encoded as %2F

How to Encode:

• When encoding a URL, the following characters should be encoded:

• Special characters (e.g., #, &, %)

• Non-ASCII characters

• Reserved characters (e.g., ?, /)

The viewport is the user's visible area of a web page. It varies with the device - it will be smaller on a
mobile phone than on a computer screen.

You should include the following <meta> element in all your web pages:

This gives the browser instructions on how to control the page's dimensions and scaling.

The width=device-width part sets the width of the page to follow the screen-width of the device
(which will vary depending on the device).

The initial-scale=1.0 part sets the initial zoom level when the page is first loaded by the browser.

HTML Iframes

An HTML <iframe> (inline frame) is used to embed another HTML document within the current HTML
document. It allows you to display content from another web page or source on your webpage. Here's
an explanation of HTML iframes:

1. Basic Syntax:

<iframe src="url"></iframe>

2. Attributes:

 Here are some commonly used attributes of the <iframe> tag in HTML5:

1. *src:* This attribute specifies the URL of the page to be displayed in the iframe.

2. *width:* It sets the width of the iframe either in pixels or as a percentage.

3. *height:* This attribute specifies the height of the iframe in pixels.

4. *frameborder:* This attribute controls the presence of a border around the iframe. It can be set to 0
to remove the border or 1 to show it.

5. *scrolling:* It determines whether the iframe should have scrollbars. It can be set to "yes" to always
show scrollbars, "no" to never show them, or "auto" to show them only when needed.

6. *sandbox:* This attribute enables a set of extra restrictions for the content within the iframe for
added security. Values can include "allow-same-origin", "allow-top-navigation", "allow-forms", and
"allow-scripts".

7. *allowfullscreen:* This attribute allows the iframe to be displayed in fullscreen mode. It is commonly
used for embedded video

8. *name:* This attribute gives the iframe a name, which can be used as the target for links or form
submissions within the iframe.

9. *loading:* This attribute specifies how the iframe content should be loaded. Possible values include
"eager" for immediate loading, "lazy" for deferred loading, and "auto" for the browser to determine the
best approach.

10. *allow:* This attribute specifies a set of permissions for the content within the iframe, such as
allowing features like camera access, geolocation, microphone access, etc. It can have values like
"camera", "microphone", "geolocation", and more.

11. *title:* This attribute provides a title for the iframe, which can be useful for accessibility purposes to
describe the contents of the iframe to screen readers or other assistive technologies.

12. *referrerpolicy:* This attribute controls how much information about the parent document should
be included in the referrer header when fetching the iframe content. Values include "no-referrer", "noreferrer-when-downgrade", "same-origin", "origin", "strict-origin", and "strict-origin-when-cross-origin".

13. *allowpaymentrequest:* This attribute allows the iframe to make use of the Payment Request API,
enabling a web application to easily request payment information from the user.

14. *csp:* This attribute specifies a Content Security Policy (CSP) for the iframe content, defining
restrictions on what resources can be loaded

15. *srcdoc:* Allows you to embed HTML content directly within the iframe using a string of HTML code
instead of an external URL.

16. *importance:* Indicates the importance of the resource to prioritize the loading of the iframe
content. Values can be "auto", "high", or "low".

4. Responsive iframes:

 To ensure iframes are responsive, you can use CSS styles like width: 100%; height: 0; padding-bottom
trick, or CSS frameworks like Bootstrap that offer responsive utilities.

It's essential to embed content responsibly with iframes, ensuring the source is trusted and secure to
prevent security risks.
